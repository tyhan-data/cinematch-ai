# Converting Categorical Data Into Numeric

```python
dataset['tag'][0]
```

```python
import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
import re
```

```python
nltk.download('stopwords')
nltk.download('wordnet')
```

```python
stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()
```

```python
def preprocess_text(text):
    # lowercase
    text = str(text).lower()
    # Remove Punctuations
    text = re.sub(r'[^a-zA-Z\s]', '', text)

    words = text.split()
    # Remove Stopwords
    words = [word for word in words if word not in stop_words]
    # Lemmatizer
    words = [lemmatizer.lemmatize(word) for word in words]

    return ' '.join(words)

dataset['tag'] = dataset['tag'].apply(preprocess_text)
```

```python
dataset['tag'][0]
```

```python
dataset = dataset.reset_index(drop = True)
```

```python
indices = pd.Series(dataset.index, index = dataset['title']).drop_duplicates()
indices.to_frame().head()
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
tfidf = TfidfVectorizer(
    max_features = 50000,
    ngram_range = (1, 2),
    stop_words = 'english'
)
```

```python
tfidf_matrix = tfidf.fit_transform(dataset['tag'])
tfidf_matrix
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
def recommend(title, n=10):
    if title not in indices:
        return ['Movie not found']

    idx = indices[title]
    sim_score = cosine_similarity(tfidf_matrix[idx], tfidf_matrix).flatten()
    sim_idx = sim_score.argsort()[::-1][1:n+1]

    return dataset['title'].iloc[sim_idx]
```

```python
recommend('Avatar').to_frame()
```

```python
import pickle

pickle.dump(tfidf_matrix, open('tfidf_matrix.pkl', 'wb'))
pickle.dump(indices, open('indices.pkl', 'wb'))
dataset.to_pickle('dataset.pkl')
pickle.dump(tfidf, open('tfidf.pkl','wb'))
```
